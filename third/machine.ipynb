```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# models
from sklearn.ensemble import RandomForestRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.ensemble import HistGradientBoostingRegressor

# RMSE testing
from sklearn.metrics import root_mean_squared_error
```

## defining training and testing data

```python
df = pd.read_csv('sdss_train_8000.csv')
df_test = pd.read_csv('sdss_validation_1000.csv')

X_train = pd.DataFrame(df, columns=['u', 'g', 'r', 'i', 'z'])
y_train = pd.DataFrame(df, columns=['redshift'])
X_test = pd.DataFrame(df_test, columns=['u', 'g', 'r', 'i', 'z'])
y_test = pd.DataFrame(df_test, columns=['redshift'])
```

## trying basic regression

```python
# random forest
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
```

```python
chosen_model = rf_model

predictions = chosen_model.predict(X_test)

rmse = root_mean_squared_error(y_test, predictions)

print(f" Root Mean Squared Error (RMSE): {rmse:.4f}")

actual_flat = np.array(y_test).ravel()
preds_flat = np.array(predictions).ravel()

preview_df = pd.DataFrame({
    'Actual Value': actual_flat,
    'Predicted Value': preds_flat,
    'Absolute Error': np.abs(actual_flat - preds_flat)
})

preview_df.head()
```

```python
# write model to file
df_test = pd.read_csv('sdss_test_without_redshift_1000.csv')
df_validation = pd.DataFrame(df_test, columns=['u', 'g', 'r', 'i', 'z'])
test_predictions = rf_model.predict(df_validation)

df['redshift'] = test_predictions
df.to_csv('baumann.csv', index=False)
```

## trying basic neural network

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)

nm_model = MLPRegressor(hidden_layer_sizes=(32, 16), max_iter=500, random_state=42, early_stopping=True)
nm_model.fit(X_scaled, y_train)
```

```python
chosen_model = nm_model 

predictions = chosen_model.predict(X_test_scaled)

rmse = root_mean_squared_error(y_test, predictions)

print(f" Root Mean Squared Error (RMSE): {rmse:.4f}")

actual_flat = np.array(y_test).ravel()
preds_flat = np.array(predictions).ravel()

preview_df = pd.DataFrame({
    'Actual Value': actual_flat,
    'Predicted Value': preds_flat,
    'Absolute Error': np.abs(actual_flat - preds_flat)
})

preview_df.head()
```

## trying high gradient boosting regressor

```python
hgb_model = HistGradientBoostingRegressor(max_iter=100, random_state=42)
hgb_model.fit(X_train, y_train)

hgb_preds = hgb_model.predict(X_test)
hgb_rmse = root_mean_squared_error(y_test, hgb_preds)

target_range = y_test.max() - y_test.min()
target_std = y_test.std()

print(f" Root Mean Squared Error (RMSE): {hgb_rmse:.4f}")

actual_flat = np.array(y_test).ravel()
preds_flat = np.array(hgb_preds).ravel()

preview_df = pd.DataFrame({
    'Actual Value': actual_flat,
    'Predicted Value': preds_flat,
    'Absolute Error': np.abs(actual_flat - preds_flat)
})

preview_df.head()
```

```python
# write model to file
df_test = pd.read_csv('sdss_test_without_redshift_1000.csv')
df_validation = pd.DataFrame(df_test, columns=['u', 'g', 'r', 'i', 'z'])
test_predictions = hgb_model.predict(df_validation)

df['redshift'] = test_predictions
df.to_csv('baumann.csv', index=False)
```

```python
import itertools
import pandas as pd
import numpy as np
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import root_mean_squared_error

def add_difference_features(df, feature_cols):
    """
    Takes a DataFrame and a list of 5 feature names.
    Returns a NEW DataFrame containing the original 5 features 
    PLUS the 10 pairwise mathematical differences.
    """

    engineered_df = df[feature_cols].copy()
    
    for col1, col2 in itertools.combinations(feature_cols, 2):
        new_col_name = f"{col1}_minus_{col2}"
        engineered_df[new_col_name] = engineered_df[col1] - engineered_df[col2]
        
    return engineered_df

FEATURES = ['u', 'g', 'r', 'i', 'z']

print("Engineering 10 new difference features...")
X_train_engineered = add_difference_features(X_train, FEATURES)
X_test_engineered = add_difference_features(X_test, FEATURES)

print(f"Original training shape: {X_train.shape}")
print(f"New engineered training shape: {X_train_engineered.shape}\n")

print("Training Gradient Boosting model on engineered features...")
diff_model = HistGradientBoostingRegressor(max_iter=100, random_state=42)
diff_model.fit(X_train_engineered, y_train)

diff_preds = diff_model.predict(X_test_engineered)
new_rmse = root_mean_squared_error(y_test, diff_preds)

print("="*45)
print(f" NEW ENGINEERED MODEL RESULTS")
print(f" RMSE with Difference Features: {new_rmse:.4f}")
print("="*45)

if hasattr(diff_model, "compute_partial_dependence"):
    print("\nTop 5 rows of your newly engineered training data:")
    display(X_train_engineered.head())
```
